```python
digits = load_digits()
X,y = digits.data, digits.target
```

```python
plt.imshow(digits.images[1], cmap='gray')
plt.title(f"Label: {digits.target[1]}")
plt.show()
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3)
model = SVC()
model.fit(X_train,y_train)
```
