```python
class Employee:

    def __init__(self,name,salary):
        self.name = name
        self.salary = salary


    def calculate_bonus(self):
        return self.salary * 0.10


emp1 = Employee("Yahya",95000)
emp2 = Employee("Khaled",84000)
emp2.calculate_bonus()
```

```python
from datetime import date

class Car:

    def __init__(self,company,model,year,engine):
        self.company = company
        self.model = model
        self.year = year
        self.engine = engine

    def stats(self):
        return f"The car you are searching is a {self.company} {self.model},manufactured in {self.year} and its engine size is {self.engine}cc"


car1 = Car("Toyota","corolla GLI", 2004, 1299)
car2 = Car("Toyota","Prius",2012,1800)
car3 = Car("Suzuki","Wagon-R",2014,660)

car1.stats()
```
